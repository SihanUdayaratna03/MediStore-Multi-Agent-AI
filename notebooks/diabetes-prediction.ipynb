```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import VotingClassifier, RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, roc_auc_score, confusion_matrix
from imblearn.over_sampling import SMOTE
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
import shap
import joblib

sns.set_style("whitegrid")
```

```python
diabetes_dataset = pd.read_csv('diabetes.csv')

cols = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
diabetes_dataset[cols] = diabetes_dataset[cols].replace(0, np.nan)
diabetes_dataset.fillna(diabetes_dataset.median(), inplace=True)

print("Data Cleaning Complete!")
diabetes_dataset.head()
```

```python
diabetes_dataset['Glucose_BMI_Ratio']   = diabetes_dataset['Glucose'] / (diabetes_dataset['BMI'] + 0.001)
diabetes_dataset['Insulin_Resistance']  = diabetes_dataset['Glucose'] * diabetes_dataset['Insulin'] / 1000
diabetes_dataset['Age_BMI_Interaction'] = diabetes_dataset['Age'] * diabetes_dataset['BMI']
diabetes_dataset['BP_Age_Risk']         = diabetes_dataset['BloodPressure'] * diabetes_dataset['Age'] / 100
diabetes_dataset['Glucose_Age']         = diabetes_dataset['Glucose'] * diabetes_dataset['Age'] / 100
diabetes_dataset['High_Glucose_Flag']   = (diabetes_dataset['Glucose'] > 125).astype(int)
diabetes_dataset['Obese_Flag']          = (diabetes_dataset['BMI'] > 30).astype(int)
diabetes_dataset['High_Risk_Age_Flag']  = (diabetes_dataset['Age'] > 45).astype(int)

print(f"Total features now: {diabetes_dataset.shape[1] - 1}")
```

```python
X = diabetes_dataset.drop('Outcome', axis=1)  
Y = diabetes_dataset['Outcome']

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, stratify=Y, random_state=42)

smote = SMOTE(random_state=42)
X_train_balanced, Y_train_balanced = smote.fit_resample(X_train, Y_train)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_balanced)
X_test_scaled  = scaler.transform(X_test)

print("✅ Data Balanced and Scaled!")
print(Y_train_balanced.value_counts())

```

```python
xgb = XGBClassifier(n_estimators=300, max_depth=6, learning_rate=0.05, random_state=42, use_label_encoder=False, eval_metric='logloss')
lgbm = LGBMClassifier(n_estimators=300, max_depth=6, learning_rate=0.05, random_state=42, verbose=-1)
rf = RandomForestClassifier(n_estimators=200, max_depth=8, random_state=42)

ensemble_model = VotingClassifier(
    estimators=[('xgb', xgb), ('lgbm', lgbm), ('rf', rf)],
    voting='soft'
)

ensemble_model.fit(X_train_scaled, Y_train_balanced)

y_pred = ensemble_model.predict(X_test_scaled)
print("Accuracy:", accuracy_score(Y_test, y_pred))
print("\nClassification Report:\n", classification_report(Y_test, y_pred))
```

```python
xgb.fit(X_train_scaled, Y_train_balanced)

explainer = shap.TreeExplainer(xgb)
shap_values = explainer.shap_values(X_test_scaled)

shap.summary_plot(shap_values, X_test_scaled, feature_names=X.columns.tolist())
```

```python
import json

joblib.dump(ensemble_model, 'diabetes_model_v2_notebook.pkl')
joblib.dump(scaler, 'scaler_v2_notebook.pkl')
joblib.dump(explainer, 'shap_explainer_notebook.pkl')

with open('feature_names_v2_notebook.json', 'w') as f:
    json.dump(X.columns.tolist(), f, indent=2)

print("Models Saved Successfully!")
```
